### Importing the Libraries

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import sklearn
import xgboost
import scipy
```

### Loading the data

```python
df = pd.read_csv("D:\others\Personal Projects 2\Practice Files July 2024\Tours and Travel Customer Churn\Customertravel.csv")
```

### Exploring the dataset

```python
## datatypes in the dataset

df.info()
```

```python
## checking for any empty values

df.isnull().sum()
```

```python
## checking for any duplicates

df.duplicated().sum()
```

```python
## dealing with duplicates

data = df.drop_duplicates()
```

```python
data.shape
```

```python
## first two rows

data.head(2)
```

### Wrangling the data

```python
## replacing the values in the hotel booking status column

data["Hotel Booking Status"] = data["Hotel Booking Status"].replace({
    "Yes" : 0,
    "No" : 1
})

data.head(2)
```

```python
data["Frequent Flyer"] = data["Frequent Flyer"].replace({
    "Yes" : 0,
    "No" : 1,
    "No Record" : 2
})

data.head(2)
```

```python
## checking the values in the annual income class

data["Annual Income Class"].unique()
```

```python
## replacing the values in the frequent flyer class

data["Annual Income Class"] = data["Annual Income Class"].replace({
    "Middle Income" : 0,
    "Low Income" : 1,
    "High Income" : 2,
})

data.head(3)
```

```python
## replacing the values in the Annual Income Class

data["Synced Social Media Accounts"] = data["Synced Social Media Accounts"].replace({
    "No" : 0,
    "Yes" : 1
})

data.head(3)
```

### Data Analysis

```python
## summary statistics

df.describe()
```

### Developing the churn prediction model

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
```

```python
## defining the dependent variable and independent variables

X = data.drop(columns = ["Churn"])
y = data["Churn"]
```

```python
## dividing into training and testing data

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size= 0.3, random_state= 5)
```

```python
# ## scaling the data

# from sklearn.preprocessing import StandardScaler

# sc = StandardScaler()

# X_train = sc.fit_transform(X_train)
# y_train = sc.fit_transform(y_train)
```

```python
## creating a regressor

lr = LogisticRegression()
```

```python
## fitting the model (training the model using the training sets)

lr.fit(X_train, y_train)
```

```python
## value of the y-intercept

lr.intercept_
```

```python
## predicting the model 

pred = lr.predict(X_test)
pred
```

```python
## calculating the accuracy 

from sklearn.metrics import confusion_matrix

confusion_matrix(y_test, pred)
```
